The course-derived results were independently checked: 101 missions, 60 Cape Canaveral missions after consolidating two naming variants, mean payload mass around 6,138 kg.

```python
from pathlib import Path
import pandas as pd,sqlite3
p=Path('Spacex.csv')
if not p.exists():
    url='https://raw.githubusercontent.com/liam-stuart/Applied-Data-Science-Capstone/main/Spacex.csv'
    data=pd.read_csv(url)
else: data=pd.read_csv(p)
con=sqlite3.connect(':memory:');data.to_sql('spacex_missions',con,index=False)
print(len(data));data.head()
```

```python
pd.read_sql_query('SELECT Launch_Site,COUNT(*) AS N FROM spacex_missions GROUP BY Launch_Site ORDER BY N DESC',con)
```

```python
pd.read_sql_query('SELECT ROUND(AVG(PAYLOAD_MASS__KG_),0) AS Mean_Payload_kg,MIN(PAYLOAD_MASS__KG_) AS Min_kg,MAX(PAYLOAD_MASS__KG_) AS Max_kg FROM spacex_missions',con)
```

```python
pd.read_sql_query('SELECT Orbit,COUNT(*) AS N FROM spacex_missions GROUP BY Orbit ORDER BY N DESC',con)
```
